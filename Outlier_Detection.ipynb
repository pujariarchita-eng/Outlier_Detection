```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Create a sample Employee Salary dataset
data = {
    "Employee": [
        "E1", "E2", "E3", "E4", "E5",
        "E6", "E7", "E8", "E9", "E10",
        "E11", "E12", "E13", "E14", "E15"
    ],
    "Salary": [
        25000, 28000, 30000, 32000, 35000,
        36000, 38000, 40000, 42000, 45000,
        48000, 50000, 52000, 55000, 200000
    ]
}

df = pd.DataFrame(data)

# 2. Display the dataset
print("EMPLOYEE SALARY DATASET")
print(df.to_string(index=False))

# 3. Calculate quartiles and IQR
Q1 = df["Salary"].quantile(0.25)
Q3 = df["Salary"].quantile(0.75)
IQR = Q3 - Q1

# 4. Calculate lower and upper bounds
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# 5. Identify potential outliers
outliers = df[
    (df["Salary"] < lower_bound) |
    (df["Salary"] > upper_bound)
]

# 6. Display statistical results
print("\nOUTLIER DETECTION RESULTS")
print(f"First Quartile (Q1): {Q1:.2f}")
print(f"Third Quartile (Q3): {Q3:.2f}")
print(f"Interquartile Range (IQR): {IQR:.2f}")
print(f"Lower Bound: {lower_bound:.2f}")
print(f"Upper Bound: {upper_bound:.2f}")

print("\nPotential Outliers:")
if outliers.empty:
    print("No potential outliers detected.")
else:
    print(outliers.to_string(index=False))

# 7. Create a box plot
plt.figure(figsize=(9, 5))
plt.boxplot(df["Salary"], patch_artist=True)
plt.title("Box Plot for Employee Salary")
plt.ylabel("Salary")
plt.xlabel("Employee Salary Distribution")
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.ticklabel_format(axis="y", style="plain")
plt.tight_layout()
plt.show()

# 8. Display findings
print("\nOBSERVATIONS")
print("1. The box plot displays the median and spread of salaries.")
print("2. Points outside the whiskers indicate potential outliers.")
print(f"3. Number of potential outliers: {len(outliers)}")

if not outliers.empty:
    print("4. Outlier salary value(s):",
          outliers["Salary"].tolist())

print("\nCONCLUSION")
print("The IQR method and box plot help identify unusually high")
print("or low salary values. Potential outliers should be checked")
print("before deciding whether to correct or remove them.")
```
